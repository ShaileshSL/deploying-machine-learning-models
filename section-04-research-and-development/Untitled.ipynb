```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

# 1. Balanced plot (4 in tall, 6 in wide)
sns.catplot(x="day", y="total_bill", kind="box",
            data=tips, height=4, aspect=1.5)

# 2. Compact square plot (3 in tall, 3 in wide)
sns.catplot(x="day", y="total_bill", kind="box",
            data=tips, height=3, aspect=1)

# 3. Wide stretched plot (5 in tall, 10 in wide)
sns.catplot(x="day", y="total_bill", kind="box",
            data=tips, height=5, aspect=2)

plt.show()
```
